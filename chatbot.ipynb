```python
conversion_data = {
    "hello": "Hi there! How can i help you today?",
    "hi": "Hello! How are you?",
    "how are you":"I'm just a big, but I'm doing great How about you",
    "what is computer vision":"computer vision is a field of AI that allows computer to see",
    "what is deep learning" : "deep learning is a subset of ai that uses neural network to",
    "bye":"Goodbye! have a nice day!",
    "thankyou":"you're welcome! happy to help",
}
```

```python
def sample_chat(user_input,chat_history=[]):
  user_input_lower = user_input.lower()
  response = None

  for key in conversion_data:
    if key in user_input_lower:
      response = conversion_data[key]
      break


  if not response:
    response="I'm sorry,I don't understand that. can you ask something else?"

    chat_history.append(("user",user_input))
    chat_history.append(("bot".response))

    return response,chat_history
```

```python
print("welcome to mini conversational AI")
print("Type 'exit' or 'quit' to end the chat.\n")
chat_history = []

while True:
  user_input = input("you:")

  if user_input.lower() in ["exit","quit"]:
    print("chat ended.Goodbye!")
    break

    bot_response, chat_history = sample_chat(user_input,chat_history)
    print("Bot:" ,bot_response)

```

```python
print("\n=======full  chat history=======")
for speaker, message in chat_history:
  print(f"{speaker}:{message}")

  useful for debugging or showing what happen during the session.
```
